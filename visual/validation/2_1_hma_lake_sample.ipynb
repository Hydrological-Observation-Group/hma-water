```python
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
path_hma_extent = "data/hma-extent/HMA/hma_gtng_202307_fill.gpkg"
path_lake_sam = "data/validation/lake_sample_2020.gpkg" 
```

```python
hma = gpd.read_file(path_hma_extent)
lake_sample = gpd.read_file(path_lake_sam)
```

```python
lake_sample.head(3)
```

```python
# 可视化展示
fig, ax = plt.subplots(figsize=(10, 8))

hma.boundary.plot(
    ax=ax, 
    edgecolor="black", 
    linewidth=0.6, 
    label="HMA boundary"
)

lake_sample.plot(
    ax=ax, 
    color="blue",       
    marker="o",       
    markersize=6,     
    label="Lake validation points",
    alpha=0.4          
)

ax.legend()
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
ax.grid(True, linestyle="--", alpha=0.4)
ax.set_title("Lake Validation Points in HMA")

plt.show()
```

```python
# 4. 保存和展示
fig.savefig("figures/validation/lake_sample.png", bbox_inches='tight', dpi=200)

```
